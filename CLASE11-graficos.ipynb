```python
# Importamos las librerías necesarias para análisis de datos y visualización
import pandas as pd          # Para manipulación de datos
import matplotlib.pyplot as plt  # Para crear gráficos
import seaborn as sns        # Para gráficos estadísticos más avanzados
import numpy as np           # Para operaciones numéricas
```

```python
# Cargamos el dataset real de autos eléctricos
data = pd.read_csv("cars-electric.csv")

print(f"Dataset: {len(data)} autos")
print("Columnas disponibles:", data.columns.tolist())
```

```python
# Creamos múltiples boxplots para diferentes variables por marca

# Configuramos subplots para mostrar múltiples boxplots
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Análisis por Marca - Múltiples Variables', fontsize=16, fontweight='bold')

# Boxplot 1: Potencia (kW) por marca
sns.boxplot(data=data, x='Make', y='(kW)', ax=axes[0,0], palette='Set2')
axes[0,0].set_title('Potencia (kW) por Marca', fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=45)
axes[0,0].grid(True, alpha=0.3)

# Boxplot 2: Autonomía (km) por marca
sns.boxplot(data=data, x='Make', y='(km)', ax=axes[0,1], palette='viridis')
axes[0,1].set_title('Autonomía (km) por Marca', fontweight='bold')
axes[0,1].tick_params(axis='x', rotation=45)
axes[0,1].grid(True, alpha=0.3)

# Boxplot 3: Consumo combinado por marca
sns.boxplot(data=data, x='Make', y='COMB (kWh/100 km)', ax=axes[1,0], palette='plasma')
axes[1,0].set_title('Consumo Combinado (kWh/100km) por Marca', fontweight='bold')
axes[1,0].tick_params(axis='x', rotation=45)
axes[1,0].grid(True, alpha=0.3)

# Boxplot 4: Tiempo de carga por marca
sns.boxplot(data=data, x='Make', y='TIME (h)', ax=axes[1,1], palette='coolwarm')
axes[1,1].set_title('Tiempo de Carga (h) por Marca', fontweight='bold')
axes[1,1].tick_params(axis='x', rotation=45)
axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Gráfico de dispersión: Marca vs Potencia (kW)
plt.figure(figsize=(12, 8))

# Creamos el scatter plot
sns.scatterplot(data=data, x='Make', y='(kW)', s=100, alpha=0.7, color='blue')

# Personalizamos el gráfico
plt.title('Gráfico de Dispersión: Potencia (kW) por Marca', fontsize=14, fontweight='bold')
plt.xlabel('Marca', fontsize=12)
plt.ylabel('Potencia (kW)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Gráfico de dispersión con seaborn
plt.figure(figsize=(12, 8))
sns.scatterplot(data=data, x='(kW)', y='(km)', hue='Make', s=60, alpha=0.7)
plt.xlabel('Potencia (kW)')
plt.ylabel('Autonomía (km)')
plt.title('Gráfico de Dispersión: Potencia vs Autonomía por Marca')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(data['(km)'], bins=15, alpha=0.7, color='lightgreen', edgecolor='black')
plt.xlabel('Autonomía (km)')
plt.ylabel('Frecuencia')
plt.title('Distribución de Autonomía en Autos Eléctricos')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(data['(kW)'], bins=20, alpha=0.7, color='skyblue', edgecolor='black')
plt.xlabel('Potencia (kW)')
plt.ylabel('Frecuencia')
plt.title('Distribución de Potencia en Autos Eléctricos')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Densidades múltiples
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

data['(kW)'].plot.density(ax=axes[0,0], alpha=0.7, color='blue')
axes[0,0].set_title('Densidad de Potencia (kW)')
axes[0,0].grid(True, alpha=0.3)

data['(km)'].plot.density(ax=axes[0,1], alpha=0.7, color='orange')
axes[0,1].set_title('Densidad de Autonomía (km)')
axes[0,1].grid(True, alpha=0.3)

data['COMB (kWh/100 km)'].plot.density(ax=axes[1,0], alpha=0.7, color='green')
axes[1,0].set_title('Densidad de Consumo')
axes[1,0].grid(True, alpha=0.3)

data['TIME (h)'].plot.density(ax=axes[1,1], alpha=0.7, color='red')
axes[1,1].set_title('Densidad de Tiempo de Carga')
axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
